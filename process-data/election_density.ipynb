```python
import pandas as pd
```

```python
elections = pd.read_csv("../data/elections_nlrb_all2024.csv")
```

```python
# how many elections are there in a week? Use the 'Election Held Date' column
elections["Election Held Date"] = pd.to_datetime(elections["Election Held Date"], errors='coerce')
elections["Week"] = elections["Election Held Date"].dt.isocalendar().week
elections["Year"] = elections["Election Held Date"].dt.year
elections.groupby(["Year", "Week"]).size().mean()
```

```python
# by fixing the employer names, how often were elections held for the same employer?
# Sort by employer and date
df_sorted = elections.sort_values(['Case Name', 'Election Held Date'])

# Compute time difference in days for each employer
df_sorted['Days_Since_Last_Election'] = df_sorted.groupby('Case Name')['Election Held Date'].diff().dt.days

# Convert to years if desired
df_sorted['Years_Since_Last_Election'] = df_sorted['Days_Since_Last_Election'] / 365
```

```python
# Average time between elections per employer
avg_years_between_elections = df_sorted.groupby('Case Name')['Days_Since_Last_Election'].mean()

# Median or other summaries
median_years_between_elections = df_sorted.groupby('Case Name')['Days_Since_Last_Election'].median()

# Overall average across all employers
overall_avg = df_sorted['Days_Since_Last_Election'].mean()
```
